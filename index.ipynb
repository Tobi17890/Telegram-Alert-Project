```python
import pandas as pd
from pathlib import Path
import requests
from html import escape
import os
from dotenv import load_dotenv


load_dotenv(override=True)

file_path = os.getenv("EXCEL_FILE_PATH")

excel_file = pd.ExcelFile(file_path)

df = pd.read_excel(file_path, sheet_name="TPP-2020")

# Fetch a customer to alert

CUSTOMER_COLUMN = "Sold-to-party Name"  
CUSTOMER_NAME = "LIM SIEK"            

matched_customer = df[
    df[CUSTOMER_COLUMN].astype(str).str.contains(CUSTOMER_NAME, case=False, na=False)
]

display(matched_customer)

customer_row = matched_customer.iloc[0]

def format_customer_message(row):
    message = "Customer Alert\n"
    message += "====================\n\n"

    for column, value in row.items():
        if pd.notna(value):
            message += f"{column}: {value}\n"

    return message

message_text = format_customer_message(customer_row)

BOT_TOKEN = os.getenv("FATHER_BOT_TOKEN")
CHAT_ID = int(os.getenv("CHAT_ID"))

def send_text_to_telegram(bot_token, chat_id, message):
    url = f"https://api.telegram.org/bot{bot_token}/sendMessage"

    payload = {
        "chat_id": chat_id,
        "text": f"<pre>{escape(message)}</pre>",
        "parse_mode": "HTML"
    }

    response = requests.post(url, data=payload)
    print(response.json())

send_text_to_telegram(BOT_TOKEN, CHAT_ID, message_text)
```
