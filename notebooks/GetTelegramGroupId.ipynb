```python
import nest_asyncio

from telethon import TelegramClient
from telethon.tl.functions.messages import (
    GetForumTopicsRequest,
)


nest_asyncio.apply()


# ============================================
# TELEGRAM ACCOUNT
# ============================================

# Keep your existing values here.
api_id = 20722579
api_hash = "45fb209e7760bb6f4dae0a3d1983a5c2"
phone_number = "+85595773695"

session_name = "my_telegram_session"


# ============================================
# GROUPS WE WANT TO CHECK
# ============================================

TARGET_GROUP_NAMES = [
    "Customer Tracking CRT&TPP R1",
    "Customer Tracking CRT&TPP R2",
    "Customer Tracking CRT&TPP R3",
    "Customer Tracking CRT&TPP R4",
    "Customer Tracking CRT&TPP R5",
    "Customer Tracking CRT&TPP R6",
]


# ============================================
# GET GROUP + TOPIC IDS
# ============================================

async def get_group_and_topic_ids():

    async with TelegramClient(
        session_name,
        api_id,
        api_hash,
    ) as client:

        await client.start(
            phone=phone_number
        )

        # limit=None makes sure we load
        # all available dialogs.
        dialogs = await client.get_dialogs(
            limit=None
        )


        # ====================================
        # STORE ALL MATCHES BY GROUP NAME
        # ====================================
        #
        # Important:
        #
        # Telegram may keep BOTH:
        #
        # Old basic group:
        #   Chat
        #   Forum=False
        #
        # New converted supergroup:
        #   Channel
        #   Forum=True
        #
        # So we must NOT keep only one dialog.
        # ====================================

        dialog_lookup = {}

        for dialog in dialogs:

            if not dialog.name:
                continue

            group_name = (
                dialog.name.strip()
            )

            if (
                group_name
                not in TARGET_GROUP_NAMES
            ):
                continue

            if (
                group_name
                not in dialog_lookup
            ):
                dialog_lookup[
                    group_name
                ] = []

            dialog_lookup[
                group_name
            ].append(
                dialog
            )


        # ====================================
        # STORE FINAL RESULTS
        # ====================================

        results = {}


        # ====================================
        # CHECK EACH REGION GROUP
        # ====================================

        for group_name in TARGET_GROUP_NAMES:

            print(
                "\n\n"
                + "=" * 75
            )

            print(
                f"SEARCHING: {group_name}"
            )

            print(
                "=" * 75
            )


            matching_dialogs = (
                dialog_lookup.get(
                    group_name,
                    [],
                )
            )


            # ================================
            # GROUP NOT FOUND
            # ================================

            if not matching_dialogs:

                print(
                    "GROUP NOT FOUND."
                )

                continue


            # ================================
            # SHOW HOW MANY MATCHES EXIST
            # ================================

            print(
                f"Matching Telegram dialogs: "
                f"{len(matching_dialogs)}"
            )


            # ================================
            # FIND VALID FORUM VERSION
            # ================================

            valid_forum_dialogs = []

            for dialog in matching_dialogs:

                entity = (
                    dialog.entity
                )

                is_forum = bool(
                    getattr(
                        entity,
                        "forum",
                        False,
                    )
                )

                is_megagroup = bool(
                    getattr(
                        entity,
                        "megagroup",
                        False,
                    )
                )

                if (
                    is_forum
                    and is_megagroup
                ):

                    valid_forum_dialogs.append(
                        dialog
                    )


            # ================================
            # NO VALID FORUM GROUP
            # ================================

            if not valid_forum_dialogs:

                print(
                    "\nNo valid Telegram "
                    "forum group was found."
                )

                print(
                    "\nAvailable matching "
                    "dialogs:"
                )

                for index, dialog in enumerate(
                    matching_dialogs,
                    start=1,
                ):

                    entity = (
                        dialog.entity
                    )

                    print(
                        "\n-------------------------"
                    )

                    print(
                        f"Match #{index}"
                    )

                    print(
                        f"ID: {dialog.id}"
                    )

                    print(
                        "Entity Type: "
                        f"{type(entity).__name__}"
                    )

                    print(
                        "Forum: "
                        f"{getattr(entity, 'forum', False)}"
                    )

                    print(
                        "Megagroup: "
                        f"{getattr(entity, 'megagroup', False)}"
                    )

                    print(
                        "Migrated To: "
                        f"{getattr(entity, 'migrated_to', None)}"
                    )

                continue


            # ================================
            # USE THE VALID FORUM GROUP
            # ================================

            target_dialog = (
                valid_forum_dialogs[0]
            )

            entity = (
                target_dialog.entity
            )


            # ================================
            # PRINT GROUP INFORMATION
            # ================================

            print(
                "\nVALID FORUM GROUP FOUND"
            )

            print(
                "-" * 75
            )

            print(
                f"GROUP: "
                f"{target_dialog.name}"
            )

            print(
                f"GROUP CHAT ID: "
                f"{target_dialog.id}"
            )

            print(
                "ENTITY TYPE: "
                f"{type(entity).__name__}"
            )

            print(
                "IS MEGAGROUP: "
                f"{getattr(entity, 'megagroup', False)}"
            )

            print(
                "TOPICS / FORUM ENABLED: "
                f"{getattr(entity, 'forum', False)}"
            )

            print(
                "-" * 75
            )


            # ================================
            # CONVERT TO INPUT ENTITY
            # ================================

            try:

                group_entity = (
                    await client.get_input_entity(
                        entity
                    )
                )

            except Exception as error:

                print(
                    "\nCould not convert "
                    "group entity:"
                )

                print(
                    type(error).__name__,
                    error,
                )

                continue


            # ================================
            # GET FORUM TOPICS
            # ================================

            try:

                topics_result = (
                    await client(
                        GetForumTopicsRequest(
                            peer=group_entity,
                            offset_date=0,
                            offset_id=0,
                            offset_topic=0,
                            limit=100,
                        )
                    )
                )

            except Exception as error:

                print(
                    "\nCould not read topics:"
                )

                print(
                    type(error).__name__
                )

                print(
                    error
                )

                continue


            # ================================
            # STORE GROUP RESULT
            # ================================

            results[
                group_name
            ] = {
                "chat_id": (
                    target_dialog.id
                ),
                "topics": {},
            }


            # ================================
            # PRINT TOPICS
            # ================================

            print(
                "\n=== TELEGRAM TOPICS ===\n"
            )

            if not topics_result.topics:

                print(
                    "No topics were returned."
                )

                continue


            for topic in topics_result.topics:

                title = getattr(
                    topic,
                    "title",
                    "Unknown",
                )

                topic_id = getattr(
                    topic,
                    "id",
                    None,
                )


                # Store it for later use.
                results[
                    group_name
                ][
                    "topics"
                ][
                    title
                ] = topic_id


                print(
                    f"{title:<35}"
                    f"Topic ID: {topic_id}"
                )


        # ====================================
        # FINAL SUMMARY
        # ====================================

        print(
            "\n\n"
            + "=" * 75
        )

        print(
            "FINAL TELEGRAM ROUTING INFORMATION"
        )

        print(
            "=" * 75
        )


        for (
            group_name,
            group_info,
        ) in results.items():

            print(
                f"\n{group_name}"
            )

            print(
                f"Chat ID: "
                f"{group_info['chat_id']}"
            )

            for (
                topic_name,
                topic_id,
            ) in group_info[
                "topics"
            ].items():

                print(
                    f"  {topic_name:<35}"
                    f"→ {topic_id}"
                )


        return results


# ============================================
# RUN IN JUPYTER
# ============================================

telegram_group_info = (
    await get_group_and_topic_ids()
)
```

```python
import nest_asyncio

from telethon import TelegramClient


nest_asyncio.apply()


api_id = 20722579
api_hash = "45fb209e7760bb6f4dae0a3d1983a5c2"
phone_number = "+85595773695"

session_name = "my_telegram_session"

TARGET_GROUP_NAME = (
    "Customer Tracking CRT&TPP R3"
)


async def check_r3():

    async with TelegramClient(
        session_name,
        api_id,
        api_hash,
    ) as client:

        await client.start(
            phone=phone_number
        )

        dialogs = await client.get_dialogs(
            limit=None
        )

        matches = []

        for dialog in dialogs:

            if not dialog.name:
                continue

            if (
                dialog.name.strip()
                == TARGET_GROUP_NAME
            ):

                matches.append(
                    dialog
                )

        print(
            f"\nFound {len(matches)} "
            "matching R3 dialog(s).\n"
        )

        for index, dialog in enumerate(
            matches,
            start=1,
        ):

            entity = dialog.entity

            print(
                "=" * 60
            )

            print(
                f"MATCH #{index}"
            )

            print(
                f"Name: {dialog.name}"
            )

            print(
                f"ID: {dialog.id}"
            )

            print(
                "Entity Type:",
                type(entity).__name__,
            )

            print(
                "Forum:",
                getattr(
                    entity,
                    "forum",
                    False,
                ),
            )

            print(
                "Megagroup:",
                getattr(
                    entity,
                    "megagroup",
                    False,
                ),
            )

            print(
                "Migrated To:",
                getattr(
                    entity,
                    "migrated_to",
                    None,
                ),
            )

            print(
                "=" * 60
            )


await check_r3()
```
