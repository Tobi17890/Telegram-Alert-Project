```python
CUSTOMER_COLUMN = "Sold-to-party Name"   # change to your real column name
CUSTOMER_NAME = "LIM SIEK"            # customer you want to alert

matched_customer = df[
    df[CUSTOMER_COLUMN].astype(str).str.contains(CUSTOMER_NAME, case=False, na=False)
]

display(matched_customer)

customer_row = matched_customer.iloc[0]
```

```python
def format_customer_message(row):
    message = "Customer Alert\n"
    message += "====================\n\n"

    for column, value in row.items():
        if pd.notna(value):
            message += f"{column}: {value}\n"

    return message

message_text = format_customer_message(customer_row)

print(message_text)

BOT_TOKEN = "8503858271:AAGZaAjVDm6r_gKV91gEVDNLXA74WKH48ts"
CHAT_ID = "1311053886"

def send_text_to_telegram(bot_token, chat_id, message):
    url = f"https://api.telegram.org/bot{bot_token}/sendMessage"

    payload = {
        "chat_id": chat_id,
        "text": f"<pre>{escape(message)}</pre>",
        "parse_mode": "HTML"
    }

    response = requests.post(url, data=payload)
    print(response.json())

send_text_to_telegram(BOT_TOKEN, CHAT_ID, message_text)
```
