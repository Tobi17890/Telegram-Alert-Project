```python
import mssql_python

server = "172.17.1.197"
database = "DataMart_CMCR_User"

connection_string = (
    f"Server={server};"
    f"Database={database};"
    "Trusted_Connection=yes;"
    "Encrypt=yes;"
    "TrustServerCertificate=yes;"
)

try:
    conn = mssql_python.connect(connection_string)
    print("Connected successfully")

except Exception as error:
    print("Connection failed:")
    print(error)
```

```python
import pandas as pd

query = """
SELECT TOP (100) *
FROM [hana].[zsd_billing_cmc_tpp];
"""

cursor = conn.cursor()
cursor.execute(query)

columns = [column[0] for column in cursor.description]
rows = cursor.fetchall()

df_tpp = pd.DataFrame(
    [tuple(row) for row in rows],
    columns=columns
)
#  count columns
print(f"Number of columns in df_tpp: {len(df_tpp.columns)}")
#  count rows
print(f"Number of rows in df_tpp: {len(df_tpp)}")

# display(df_tpp)
```
